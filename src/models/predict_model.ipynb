## 6. PREPARING INPUT DATA

### 6.3 Check the Input Feature Columns

## 8. PREDICTION PROBABILITIES

Prediction probabilities represent the model's estimated probability for each possible performance class.
Not every machine learning classifier provides probability estimates. Therefore, this section checks whether the trained model supports predict_proba() before calculating probabilities.


### 8.1 Check Whether the Trained Model Supports Prediction Probabilites

```python
# Check whether the trained model supports prediction probabilities

if hasattr(trained_model, "predict_proba"):

    prediction_probabilities = trained_model.predict_proba(
        prediction_input
    )

    print(f"Prediction probabilities generated successfully.")
    print(f"Probability matrix shape: {prediction_probabilities.shape}")

else:

    prediction_probabilities = None

    print(f"The selected trained model does not provide prediction probabilities.")
```

### 8.2 Extract Model Classes When Probability Estimates are Available

```python
# Extract model classes when probability estimates are available

if prediction_probabilities is not None:

    model_classes = trained_model.classes_

    print(f"Performance classes available in the model:")

    for index, class_value in enumerate(model_classes):

        print(f"Class {index + 1}: {class_value}")

else:

    model_classes = None

    print(f"No probability classes available.")
```

### 8.3 Add Probabilites Columns to the Prediction Results

```python
# Add probability columns to the prediction results

if prediction_probabilities is not None:

    for index, class_value in enumerate(model_classes):

        probability_column = (
            f"Probability_Performance_{class_value}"
        )

        prediction_results[probability_column] = (
            prediction_probabilities[:, index]
        )

        print(
            f"Added probability column: "
            f"{probability_column}"
        )

else:

    print(f"No probability columns were added.")
```

### Insight

Probability values provide additional information beyond the predicted class. For example, a prediction with a probability close to 0.95 indicates stronger model confidence than a prediction with a probability close to 0.55.
However, model probability should be interpreted as a model estimate rather than a guarantee of employee performance.


## 9. EMPLOYEE-WISE PREDICTION RESULTS

Employee-wise prediction results combine the available employee information with the model's predicted performance rating.
An employee identifier is used when available so that predictions can be traced back to individual records.


### 9.1 Identify a Suitable Employee Identifier

### 10. IDENTIFICATION OF POTENTIAL LOW PERFORMERS

One important business application of an employee performance prediction model is to identify employees whose predicted performance rating falls into a lower category.
This should not be interpreted as a final judgment about an employee. Instead, these records can be treated as cases for additional review, support, training, or development.


### 10.1 Identify The Lowest Predicted Performance Rating

```python
# Identify the lowest predicted performance rating

unique_predictions = sorted(
    prediction_results[
        "PredictedPerformanceRating"
    ].dropna().unique()
)

print(f"Unique predicted performance ratings: {unique_predictions}")
```

### 10.2 Define The Potential Low-Performance Threshold

```python
# Define the potential low-performance threshold

if len(unique_predictions) > 1:

    low_performance_threshold = unique_predictions[0]

else:

    low_performance_threshold = None

print(
    f"Potential low-performance threshold: "
    f"{low_performance_threshold}"
)
```

### 10.3 Identify Employees Predicted in the Lowest Performance Category

```python
# Identify employees predicted in the lowest performance category

if low_performance_threshold is not None:

    potential_low_performers = prediction_results[
        prediction_results[
            "PredictedPerformanceRating"
        ] == low_performance_threshold
    ].copy()

    print(
        f"Number of employees predicted in the lowest "
        f"performance category: "
        f"{len(potential_low_performers)}"
    )

else:

    potential_low_performers = pd.DataFrame()

    print(
        f"Potential low performers could not be identified "
        f"because only one prediction class is available."
    )
```

### 10.4 Calculate Percentage of Potential Low performers

```python
# Calculate percentage of potential low performers

if len(prediction_results) > 0:

    low_performer_percentage = (
        len(potential_low_performers)
        / len(prediction_results)
        * 100
    )

else:

    low_performer_percentage = 0

print(
    f"Potential low performers: "
    f"{len(potential_low_performers)} employees"
)

print(
    f"Percentage of potential low performers: "
    f"{low_performer_percentage:.2f}%"
)
```

### 10.5 Display Potential Lower Performers

```python
# Display potential low performers

if not potential_low_performers.empty:

    if employee_id_column is not None:

        low_performer_view = potential_low_performers[
            [
                employee_id_column,
                "PredictedPerformanceRating"
            ]
        ]

    else:

        low_performer_view = potential_low_performers[
            [
                "PredictedPerformanceRating"
            ]
        ]

    print(f"Potential low performer records:")

    display(
        low_performer_view.head(20)
    )

else:

    print(f"No potential low performer records identified.")
```

### Insight

The model identifies employees belonging to the lowest predicted performance category. These employees should be considered for supportive interventions rather than automatic negative decisions.

## 11. PREDICTION INTERPRETATION

The prediction output should be interpreted as a machine learning estimate based on historical patterns in the training data.
A predicted performance rating does not directly explain why an employee received that prediction. It represents the class selected by the trained model from the employee's available characteristics.


### 11.1 Calculate Overall Prediction Statistics

```python
# Calculate overall prediction statistics

total_employees = len(prediction_results)

prediction_counts = (
    prediction_results[
        "PredictedPerformanceRating"
    ]
    .value_counts()
    .sort_index()
)

print(f"Total employees evaluated: {total_employees}")
print(f"Number of predicted performance categories: {len(prediction_counts)}")

for rating, count in prediction_counts.items():

    percentage = (
        count / total_employees * 100
        if total_employees > 0
        else 0
    )

    print(
        f"Performance Rating {rating}: "
        f"{count} employees "
        f"({percentage:.2f}%)"
    )
```

### Insight

The distribution of predicted performance ratings provides an overall view of how the model classifies the employee population.
The predictions should be interpreted together with model evaluation metrics from the training stage.

## 15. MODEL LIMITATIONS

The prediction model has several limitations.

First, the model can only learn patterns represented in the training data. If the historical data contains bias, incomplete information, or measurement problems, those issues can influence future predictions.

Second, a prediction does not establish causation. A feature that is strongly associated with a model prediction does not necessarily cause employee performance to increase or decrease.

Third, employee behavior and organizational conditions can change over time, meaning that a model trained on historical data may become less accurate if it is not periodically evaluated and retrained.


### 15.1 Display Model Limitations

```python
# Display model limitations

model_limitations = [
    "Predictions depend on the quality of the training data.",
    "Historical patterns may not represent future employee behavior.",
    "Model predictions do not establish causal relationships.",
    "Prediction errors are possible.",
    "Important employee factors may not be present in the dataset.",
    "Model performance can change when business conditions change.",
    "The model should be periodically validated and retrained."
]

print(f"Model limitations:")

for index, limitation in enumerate(
    model_limitations,
    start=1
):

    print(f"{index}. {limitation}")
```

### Insight

The prediction model should be treated as a decision-support system with measurable uncertainty rather than as a perfect employee evaluation mechanism.
