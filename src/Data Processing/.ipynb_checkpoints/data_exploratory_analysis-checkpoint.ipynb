## 17. OUTLIERS ANALYSIS

Outliers are observations that are unusually high or low compared with the rest of the data. They can influence statistical summaries and some machine learning algorithms.

The Interquartile Range (IQR) method is used to identify potential outliers in numerical variables. An observation is considered a potential outlier when it falls below `Q1 - 1.5 x IQR or above Q3 + 1.5 x IQR.`

### 17.1 Check Dataset Size Before Outliers Removal

```python
# Check dataset before outlier removal

before_rows = len(df)
before_columns = len(df.columns)

print("Before Outlier Removal:\n")
print(f"Number of Rows: {before_rows}")
print(f"Number of Columns: {before_columns}")
```

### 17.2 Calculate Outliers Using The IQR Method

```python
# Calculate outliers using the IQR method

outlier_summary = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_percentage = (
        outlier_count / len(df)
    ) * 100

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outlier_count,
        "Outlier_Percentage": outlier_percentage
    })

# Convert list to DataFrame AFTER the loop
outlier_summary = pd.DataFrame(outlier_summary)

# Display the final summary
display(outlier_summary)
```

### 17.3 Remove Outliers - Remove Outliers Using The IQR Method

```python
# Remove outliers using the IQR method

df_clean = df.copy()

for column in numerical_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean = df_clean[
    (df_clean[column] >= lower_bound) &
    (df_clean[column] <= upper_bound)
        ]
print("Outliers removal completed.")
```

### 17.4 After Outlier Removal - Check Dataset After Outliers Removal

```python
# Check dataset after outlier removal

after_rows = len(df_clean)
after_columns = len(df_clean.columns)

print("After Outlier Removal")
print(f"Number of rows: {after_rows}")
print(f"Number of columns: {after_columns}")
```

### 17.5 Before VS After Comparison - Compare Dataset Before and After Outlier Removal

```python
# Compare dataset before and after outlier removal

rows_removed = before_rows - after_rows
removal_percentage = (rows_removed / before_rows) * 100

comparison = pd.DataFrame({
    "Metric": [
        "Number of Rows",
        "Number of Columns",
        "Rows Removed",
        "Removal Percenatge"
    ],
    "before Outlier Removal": [
        before_rows,
        before_columns,
        "-",
        "-"
    ],
    "After Outlier Removal": [
        after_rows,
        after_columns,
        rows_removed,
        f"{removal_percentage:.2f}%"
    ]
})
display(comparison)
        
```
