### 6. DATA QUALITY ASSESSMENT

Data quality assessment is performed to identify potential problems in the dataset before analysis. Important checks including the number of records, number of column, missing values, duplicate records, unquie values, and data types.

Thesse checks helpd determine whether the dataste is complete, consistent, and suitable for futher preprocessing.

### 6.1 Create a Data Quality Summary

```python
# Create a data quality summary

quality_summary = pd.DataFrame({
    "Columns": df_processed.columns,
    "Data_Type": df_processed.dtypes.astype(str).values,
    "Non_Null_Count": df_processed.notnull().sum().values,
    "Missing_Count": df_processed.isnull().sum().values,
    "Unique_Count": df_processed.nunique().values
})

quality_summary["Missing_Percentage"]=(
    quality_summary["Missing_Count"]
    / len(df_processed)
)*100

display(quality_summary)

print(f"Total records assessed: {len(df_processed)}\n")
print(f"Total columns assessed: {df_processed.shape[1]}\n")
print(f"Total missing values in dataset:"
      f"{df_processed.isnull().sum().sum()}\n")
print(f"Total duplicate records:"
      f"{df_processed.duplicated().sum()}")
```

### Insight

The data quality assessment provides a cloumn-level view of completeness and uniqueness. Columns with missing values, unexpected data types, or unusually low/high numbers of unique values should be investigated before the dataset is used for modeling.

### 7. MISSING VALUE ANALYSIS

Missing values can affect statistical analysis and machine learning algorithms. Therfore, every column is checked for missing observations. The number and percentage of missing values are calculated to determine whether additional treatment is required.

The treatment strategy depends on the amount of missing data and the meaning of the corresponding variable. Missing values should not automatically be removed without understanding their impact.

### 7.1 Calculate Missing Values

```python
# Calculate missing values

missing_summary = pd.DataFrame({
    "Missing_Count": df_processed.isnull().sum(),
    "Missing_Precentage": (
        df_processed.isnull().sum()
        / len(df_processed)
    )*100
})

missing_summary = missing_summary.sort_values(
    by = "Missing_Count",
    ascending = False
)       
```

### 7.2 Identify Columns Containing Missing Values

```python
# Identify columns containing missing values

missing_columns = missing_summary[
    missing_summary["Missing_Count"]>0
]

print(f"Number of columns containing missing values: {len(missing_columns)}")

if len(missing_columns)>0:

    print("Columns with missing values:\n")

    for column in missing_columns.index:
        count = missing_summary.loc[column, "Missing_Count"]
        percentage = missing_summary.loc[column, "Missing_Percentage"]

        print(
        f"{column}:"
        f"{int(count)} missing values"
        f"({percentage:.2f}%)"
        )
else:
        print("No missing values were found in the dataset.")
```

### Insight

The missing-value analysis determines whether the dataset requires imputation or record removal. If no missing values are present, no-missing-value treatment is necessary. If missing values exist, the approprite treatment should be selected based on the variable type and business meaning.

### 8. DUPLICATE RECORDS ANALYSIS

Duplicate records can create bias by causing the same employee observation to appear more than once. Therefore, the dataset is checked for completely duplicated rows.

If duplicate records are found, they can normally be removed when they represent accidental duplication rather than legitimate repeated observation.

### 8.1 Count Duplicate Records


```python
# Count duplicate records

duplicate_count = df_processed.duplicated().sum()

print(f"Number of duplicate records: {duplicate_count}")
```

### 8.2 Display Duplicate Records If Present 

```python
# Display duplicate records if present

if duplicate_count > 0:

    duplicate_rows = df.processed[
        df_processed.duplicated(keep = False)
    ]

    print(f"Displaying {len(duplicate_rows)} duplicate-related rows:")
    display(duplicate_rows)
else:
    print("No duplicate records found.")
```

### 8.3 Remove Duplicate Records


```python
# Remove duplicate records

before_duplicates = len(df_processed)
df_processed = df_processed.drop_duplicates()

after_duplicates = len(df_processed)

removed_duplicates = (
    before_duplicates - after_duplicates
)

print(f"Records before duplicate removal: {before_duplicates}\n")
print(f"Records after duplicate removal: {after_duplicates}\n")
print(f"Duplicate records removed: {removed_duplicates}")
```

### Insight

Duplicate records have been checked and removed when present. Removing accidental duplicates prevents repeadted observations from dispropoortiontely influencing descriptive statistics and machine learning models.

### 9. TARGET VARIABLE VALIDATION

The target variable is te variable that the machine elarning model will attempt to project. For the INX Future Inc Employee Performance project, the expected target variable is `PerformanceRating`.

The target variable must be checked for existence, data type, missing values, unique values, and class distribution. This validation ensures that the target is correctly identified before feature engineering and model training.

### 9.1 Define Target Variable

```python
target_column = "PerformanceRating"

print(f"Target variable selected: {target_column}")

if target_column in df_processed.columns:
    print("Target variable exists in the dataset.")
else:
    print(
        f"WARNING: Target variable '{target_column}'"
        f"was not fount in the dataset."
    )
```

### 9.2 Validate Target Variable 

```python
# Validate target variable

if target_column in df_processed.columns:
    print(f"Target data type: {df_processed[target_column].dtype}\n")
    
    print(
    f"Missing target values: "
    f"{df_processed[target_column].isnull().sum()}\n"
    )

    print(
    f"Number of unique target values: "
    f"{df_processed[target_column].nunique()}\n"
    )

    print(
    f"Unique target values: "
    f"{sorted(df_processed[target_column].dropna().unique())}"
    )
```

### 9.3 Target Distribution - Display Target Variable Distribution

```python
# Display target variable distribution

if target_column in df_processed.columns:

    target_distribution = (
        df_processed[target_column]
        .value_counts()
        .sort_index()
    )

    display(target_distribution.to_frame("Count"))
```

### 9.4 Calculate Target Percentage Distribution


```python
# Calculate target percentage distribution

if target_column in df_processed.columns:
    target_percentage = (
        df_processed[target_column]
        .value_counts(normalize = True)
        .sort_index()
        * 100
    )

    for value, percentage in target_percentage.items():

        print(
            f"PerformanceRating = {value}:"
            f"{percentage:.2f}%"
        )
```

### Insight

The target variable is validate for availability, completeness, and class distribution. Understanding the target distribution is important because a highly imbalanced target may require special consideration during model evaluation and training.
